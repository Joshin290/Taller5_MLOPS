```python
import sklearn
import pandas as pd
import mlflow
import numpy as np
import pandas as pd
import mlflow.sklearn

from itertools import product
from sqlalchemy import create_engine

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import precision_score, recall_score, f1_score, classification_report

from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder
)

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.metrics import (
    accuracy_score,
    classification_report
)

mlflow.sklearn.autolog(disable=True)
print("scikit-learn:", sklearn.__version__)
print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)
print("MLflow:", mlflow.__version__)
```

```python
# ============================================================
# 2. CONFIGURAR MLFLOW
# ============================================================

mlflow.set_tracking_uri(
    "http://mlflow:5000"
)

mlflow.set_experiment(
    "Penguins_Classification_RF"
)

print()
print("============================================")
print("MLFLOW CONFIGURADO")
print("============================================")
print("Tracking URI: http://mlflow:5000")
print("Experimento: Penguins_Classification_RF")


# ============================================================
# 3. CONECTAR A POSTGRESQL
# ============================================================

engine = create_engine(
    "postgresql+psycopg2://admin:supersecret@data_db:5432/penguinsdb"
)

df = pd.read_sql(
    "SELECT * FROM penguins",
    engine
)

print()
print("============================================")
print("DATOS CARGADOS")
print("============================================")
print(f"Cantidad de registros: {len(df)}")
print(f"Cantidad de columnas: {len(df.columns)}")
print()
print("Columnas:")
print(list(df.columns))

display(df.head())


# ============================================================
# 4. LIMPIEZA
# ============================================================

df = df.dropna(
    subset=["species"]
)

X = df.drop(
    "species",
    axis=1
)

y = df["species"]

print()
print("============================================")
print("LIMPIEZA DE DATOS")
print("============================================")
print(f"Registros después de limpieza: {len(df)}")
print()
print("Distribución de clases:")
print(y.value_counts())


# ============================================================
# 5. CARACTERÍSTICAS
# ============================================================

numeric_features = [
    "bill_length_mm",
    "bill_depth_mm",
    "flipper_length_mm",
    "body_mass_g"
]

categorical_features = [
    "island",
    "sex"
]

print()
print("============================================")
print("CARACTERÍSTICAS")
print("============================================")
print("Variables numéricas:")
print(numeric_features)
print()
print("Variables categóricas:")
print(categorical_features)


# ============================================================
# 6. PREPROCESAMIENTO
# ============================================================

numeric_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore"
            )
        )
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer,
            numeric_features
        ),
        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ]
)

print()
print("============================================")
print("PREPROCESAMIENTO CONFIGURADO")
print("============================================")


# ============================================================
# 7. TRAIN / VALIDATION / TEST
# ============================================================

X_temp, X_test, y_temp, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

X_train, X_val, y_train, y_val = train_test_split(
    X_temp,
    y_temp,
    test_size=0.25,
    random_state=42,
    stratify=y_temp
)

print()
print("============================================")
print("DIVISIÓN DE DATOS")
print("============================================")
print(f"Train:      {len(X_train)} registros")
print(f"Validation: {len(X_val)} registros")
print(f"Test:       {len(X_test)} registros")
print()
print("Train:      60%")
print("Validation: 20%")
print("Test:       20%")


# ============================================================
# 8. GRID DE 30 COMBINACIONES
# ============================================================

param_grid = {
    "n_estimators": [
        50,
        100,
        150
    ],

    "max_depth": [
        None,
        10
    ],

    "min_samples_split": [
        2,
        5,
        10,
        15,
        20
    ]
}

combinations = list(
    product(
        param_grid["n_estimators"],
        param_grid["max_depth"],
        param_grid["min_samples_split"]
    )
)

print()
print("============================================")
print("GRID DE HIPERPARÁMETROS")
print("============================================")
print(f"Total de combinaciones: {len(combinations)}")


# ============================================================
# 9. VARIABLES PARA RESULTADOS
# ============================================================

results = []

best_f1 = -1
best_model = None
best_params = None
best_run_id = None


# ============================================================
# 10. ENTRENAR LAS 30 COMBINACIONES
# ============================================================

for i, (
    n_estimators,
    max_depth,
    min_samples_split
) in enumerate(
    combinations,
    start=1
):

    print()
    print("============================================")
    print(f"COMBINACIÓN {i}/30")
    print("============================================")

    print(
        f"n_estimators:      {n_estimators}"
    )

    print(
        f"max_depth:         {max_depth}"
    )

    print(
        f"min_samples_split: {min_samples_split}"
    )

    # --------------------------------------------------------
    # Crear modelo
    # --------------------------------------------------------

    model = Pipeline(
        steps=[
            (
                "preprocessor",
                preprocessor
            ),

            (
                "classifier",
                RandomForestClassifier(
                    n_estimators=n_estimators,
                    max_depth=max_depth,
                    min_samples_split=min_samples_split,
                    random_state=42
                )
            )
        ]
    )

    # --------------------------------------------------------
    # Crear run de MLflow
    # --------------------------------------------------------

    with mlflow.start_run(
        run_name=f"RF_Combination_{i}"
    ) as run:

        # Parámetros
        mlflow.log_param(
            "n_estimators",
            n_estimators
        )

        mlflow.log_param(
            "max_depth",
            max_depth
        )

        mlflow.log_param(
            "min_samples_split",
            min_samples_split
        )

        # ----------------------------------------------------
        # Entrenamiento
        # ----------------------------------------------------

        model.fit(
            X_train,
            y_train
        )

        # ----------------------------------------------------
        # Validación
        # ----------------------------------------------------

        y_val_pred = model.predict(
            X_val
        )

        val_precision = precision_score(
            y_val,
            y_val_pred,
            average="weighted",
            zero_division=0
        )

        val_recall = recall_score(
            y_val,
            y_val_pred,
            average="weighted",
            zero_division=0
        )

        val_f1 = f1_score(
            y_val,
            y_val_pred,
            average="weighted",
            zero_division=0
        )

        # Registrar métricas en MLflow
        mlflow.log_metric(
            "validation_precision",
            val_precision
        )
        mlflow.log_metric(
            "validation_recall",
            val_recall
        )
        mlflow.log_metric(
            "validation_f1_score",
            val_f1
        )

        # ----------------------------------------------------
        # Guardar resultado
        # ----------------------------------------------------

        results.append(
            {
                "Combinación": i,
                "n_estimators": n_estimators,
                "max_depth": max_depth,
                "min_samples_split": min_samples_split,
                "Validation Precision": val_precision,
                "Validation Recall": val_recall,
                "Validation F1-Score": val_f1,
                "Run ID": run.info.run_id
            }
        )

        print()
        print(
            f"Validation F1-Score: {val_f1:.4f} | Precision: {val_precision:.4f} | Recall: {val_recall:.4f}"
        )

        # ----------------------------------------------------
        # Determinar mejor modelo (optimizando por F1-Score)
        # ----------------------------------------------------

        if val_f1 > best_f1:

            best_f1 = val_f1

            best_model = model

            best_params = {
                "n_estimators": n_estimators,
                "max_depth": max_depth,
                "min_samples_split": min_samples_split
            }

            best_run_id = run.info.run_id

            print()
            print("*** NUEVO MEJOR MODELO ***")
            print(
                f"F1-Score: "
                f"{best_f1:.4f}"
            )

            print(
                f"Run ID: "
                f"{best_run_id}"
            )


# ============================================================
# 11. TABLA ÚNICA DE LAS 30 COMBINACIONES
# ============================================================

results_df = pd.DataFrame(
    results
)

# Ordenar por F1-Score descendente
results_df = results_df.sort_values(
    by="Validation F1-Score",
    ascending=False
).reset_index(
    drop=True
)

# Identificar todas las combinaciones que empatan
results_df["Mejor modelo"] = (
    results_df["Validation F1-Score"]
    == best_f1
)

print()
print()
print("============================================")
print("RESULTADOS DE LAS 30 COMBINACIONES")
print("============================================")

display(
    results_df
)


# ============================================================
# 12. EVALUACIÓN DEL MEJOR MODELO EN TEST
# ============================================================

y_test_pred = best_model.predict(
    X_test
)

test_precision = precision_score(
    y_test,
    y_test_pred,
    average="weighted",
    zero_division=0
)

test_recall = recall_score(
    y_test,
    y_test_pred,
    average="weighted",
    zero_division=0
)

test_f1 = f1_score(
    y_test,
    y_test_pred,
    average="weighted",
    zero_division=0
)

print()
print("============================================")
print("EVALUACIÓN DEL MEJOR MODELO")
print("============================================")

print(
    f"Validation F1-Score: "
    f"{best_f1:.4f}"
)

print(
    f"Test F1-Score: "
    f"{test_f1:.4f}"
)

print(
    f"Test Precision: "
    f"{test_precision:.4f}"
)

print(
    f"Test Recall: "
    f"{test_recall:.4f}"
)

print()
print("Reporte de clasificación:")
print(
    classification_report(
        y_test,
        y_test_pred
    )
)


# ============================================================
# 13. REGISTRAR RESULTADOS DEL MODELO FINAL
# ============================================================

print()
print("============================================")
print("REGISTRANDO MEJOR MODELO")
print("============================================")

with mlflow.start_run(
    run_name="Best_RandomForest"
) as best_run:

    # Registrar hiperparámetros
    mlflow.log_params(
        best_params
    )

    # Registrar métricas
    mlflow.log_metric(
        "validation_f1_score",
        best_f1
    )

    mlflow.log_metric(
        "test_f1_score",
        test_f1
    )

    mlflow.log_metric(
        "test_precision",
        test_precision
    )

    mlflow.log_metric(
        "test_recall",
        test_recall
    )

    # Registrar información
    mlflow.set_tag(
        "best_experiment_run_id",
        best_run_id
    )

    mlflow.set_tag(
        "model_selection",
        "Best validation f1_score"
    )

    mlflow.set_tag(
        "total_combinations",
        "30"
    )

    mlflow.set_tag(
        "train_validation_test_split",
        "60/20/20"
    )

    # --------------------------------------------------------
    # REGISTRAR MODELO
    # --------------------------------------------------------

    mlflow.sklearn.log_model(
    sk_model=best_model,
    artifact_path="model",
    registered_model_name="Penguins_RF"
    )

    best_model_run_id = best_run.info.run_id


# ============================================================
# 14. TABLA FINAL RESUMEN
# ============================================================

summary_df = pd.DataFrame(
    [
        {
            "Mejor combinación": (
                results_df.loc[
                    results_df["Validation F1-Score"]
                    == best_f1,
                    "Combinación"
                ].iloc[0]
            ),

            "n_estimators": best_params[
                "n_estimators"
            ],

            "max_depth": best_params[
                "max_depth"
            ],

            "min_samples_split": best_params[
                "min_samples_split"
            ],

            "Validation F1-Score": best_f1,

            "Test F1-Score": test_f1,

            "Test Precision": test_precision,

            "Test Recall": test_recall,

            "Run entrenamiento": best_run_id,

            "Run modelo registrado": best_model_run_id,

            "Modelo MLflow": "Penguins_RF"
        }
    ]
)

print()
print()
print("============================================")
print("RESUMEN FINAL")
print("============================================")

display(
    summary_df
)

print()
print("============================================")
print("ENTRENAMIENTO FINALIZADO CORRECTAMENTE")
print("============================================")
```
